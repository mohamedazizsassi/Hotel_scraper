```python
# Merge the three per-model comparison.json outputs into one comparison.json.
# Attach the output datasets from notebooks 21, 22, 23 as inputs before running.
import glob, json

merged = {}
for model in ("lightgbm", "catboost", "xgboost"):
    # Each notebook writes /kaggle/working/bakeoff/comparison.json with 1 key
    hits = sorted(glob.glob(f"/kaggle/input/**/comparison.json", recursive=True))
    for h in hits:
        data = json.loads(open(h).read())
        for k, v in data.items():
            if k not in merged:
                merged[k] = v
                print(f"loaded {k} from {h}")

print("
=== merged report ===")
print(json.dumps(merged, indent=2))

import os
os.makedirs("/kaggle/working/bakeoff", exist_ok=True)
with open("/kaggle/working/bakeoff/comparison.json", "w") as f:
    json.dump(merged, f, indent=2)
print("
wrote /kaggle/working/bakeoff/comparison.json")
```
